## 0. 지난 시간 복습 · 슬라이드 2
지난 시간 마지막 셀입니다. 기록 한 줄을 나누고, 숫자로 바꾸고, 비교해서 **조건의 결과(True / False)를 출력**했습니다.
오늘은 여러 값을 담는 자료형을 다듬은 뒤, 조건의 결과에 따라 **실행할 코드를 고르는** 데까지 갑니다.

```python
# [0-1] 지난 시간 종합: 기록 한 줄 → 나누기 → 변환 → 비교 → 출력
line = "2026-09-30 14:00,T01,29.1,85"
parts = line.split(",")
t = float(parts[2])
rh = int(parts[3])
vent = t >= 28 or rh >= 85
print(f"{parts[1]} {t:.1f} ℃ / {rh} % → 환기 필요: {vent}")
```

## 1. 리스트 — 추가 · 삭제 · 정렬 · 통계 · 실습 1 · 슬라이드 4–5
지난 시간에 리스트의 생김새와 값 꺼내기를 봤습니다. 실제 데이터를 다룰 때 하는 일은 세 가지입니다. **추가, 삭제, 정렬.** 그리고 통계 함수 네 개를 더합니다.

```python
# [1-1] 추가: append(값) — 맨 뒤에 붙인다
temps = [17.5, 19.8, 25.4, 26.9, 22.1]    # 06·09·12·15·18시 온도 (℃)
temps.append(29.5)                       # 21시 측정값
print(temps)
print(len(temps))
```

```python
# [1-2] 삭제: del 리스트[번호], pop() — 마지막 값을 꺼내면서 삭제
temps = [17.5, -99.0, 25.4, 26.9, 22.1, 20.3]   # 09시 값이 결측 코드 -99.0
del temps[1]                                     # 번호로 삭제
print(temps)
last = temps.pop()                               # 마지막 값을 꺼내면서 삭제
print(last, temps)
```

**예상해 보기** [1-2]를 한 번 더 실행하면? (리스트를 다시 만드는 줄이 있으므로 결과는 같습니다. 그 줄을 지우고 실행하면 어떻게 될까요?)

```python
# [1-3] 정렬: sort()는 리스트 자체를 바꾼다, sorted()는 정렬된 새 리스트를 돌려준다
temps = [17.5, 19.8, 25.4, 26.9, 22.1]
temps.sort()                            # 오름차순, temps 자체가 바뀐다
print(temps)
temps.sort(reverse=True)                # 내림차순
print(temps)
original = [17.5, 19.8, 25.4, 26.9, 22.1]
ranked = sorted(original)               # 원본은 그대로, 새 리스트
print(original, ranked)
```

### 심화 주의 — sort()와 sorted()
- `temps.sort()`는 **값을 돌려주지 않고** 리스트를 제자리에서 바꿉니다. `x = temps.sort()`라고 쓰면 `x`는 `None`입니다.
- 원본 순서를 보존해야 하면 `sorted(temps)`로 새 리스트를 받습니다. 지난 시간 "변수와 객체"의 원본 보존 문제와 같은 맥락입니다.

```python
# [1-4] 통계: min · max · sum · len — 리스트를 넣으면 값 하나를 돌려준다
temps = [17.5, 19.8, 25.4, 26.9, 22.1]
t_max = max(temps)
t_min = min(temps)
t_mean = sum(temps) / len(temps)
print(f"최고 {t_max} 최저 {t_min} 일교차 {t_max - t_min:.1f}")
print(f"평균 {t_mean:.1f} ℃")
```

```python
# [1-5] 생육도일(GDD, growing degree days): 지난 시간 식에 max · min을 넣는다
t_base = 10.0
gdd = (max(temps)+min(temps))/2 - t_base
print(f"생육도일 {gdd:.1f}")
```

## 2. 딕셔너리 — 수정 · 조회 · 묶음 · 실습 2 · 슬라이드 6–7
한 관측값에 이름을 붙여 두면 리스트보다 읽기 쉽습니다. 지난 시간에 생김새와 값 하나 꺼내기를 봤으니, 오늘은 항목을 **추가 · 수정 · 삭제**하고 **안전하게 조회**합니다.

```python
# [2-1] 추가와 수정: 딕셔너리[키] = 값 — 없는 키면 추가, 있는 키면 수정
sensor = {"id": "T01", "temp": 23.6}
sensor["rh"] = 68            # 추가
sensor["temp"] = 24.2        # 수정
print(sensor)
```

```python
# [2-2] 삭제: del 딕셔너리[키]
del sensor["rh"]
print(sensor)
```

없는 키를 대괄호로 꺼내면 오류가 납니다. 실행해서 마지막 줄을 읽어 봅니다.

```python
# [2-3] 없는 키를 대괄호로 꺼내면 → KeyError
sensor["co2"]
```

```python
# [2-4] 안전한 조회: get(키, 기본값), in은 키가 있는지
print(sensor.get("co2"))            # 없으면 None
print(sensor.get("co2", 0))        # 없으면 기본값
print("temp" in sensor, "co2" in sensor)
```

```python
# [2-5] 키 · 값 · 쌍 목록 — 이런 기능이 있다. 반복문에서 쓰게 된다 (다음 시간)
print(sensor.keys())
print(sensor.values())
print(sensor.items())
```

## 3. 문자열 정리 · 자료형 선택 기준 · 실습 3 · 슬라이드 8–10
로거나 파일에서 온 문자열에는 공백, 줄바꿈, 단위 글자가 섞여 있습니다. 정리 흐름은 **원본 → 불필요한 부분 제거 → 나누기 → 필요하면 다시 합치기**입니다.

```python
# [3-1] 앞뒤 공백 · 줄바꿈 제거 → strip(), 쉼표로 나누기 → split()
line = "  T01,29.1,85\n"
print(repr(line))               # repr: 공백과 줄바꿈까지 그대로 보여 준다
clean = line.strip()
parts = clean.split(',')
print(parts)
```

```python
# [3-2] 글자 바꾸기 · 지우기 → replace(바꿀 것, 새 것)
value = "23.6C"
t = float(value.replace("C", ""))      # 단위 글자를 지운 뒤 숫자로
print(t + 0.5)
print("T01,29.1,85".replace(",", ";"))   # 구분자 바꾸기
```

```python
# [3-3] 다시 한 줄로 → "구분자".join(문자열 리스트)
fields = ["T01", "29.1", "85"]
print(",".join(fields))
print(" | ".join(fields))
```

### 자료형 선택 기준 · 슬라이드 9

| 자료형 | 모양 | 순서 | 변경 | 꺼내기 | 쓰임 |
|---|---|---|---|---|---|
| 문자열 `str` | `"…"` | 있음 | 불가 | 번호 | 글자 데이터, 기록 한 줄 |
| 리스트 `list` | `[ , ]` | 있음 | 가능 | 번호 | 측정값 목록 |
| 튜플 `tuple` | `( , )` | 있음 | 불가 | 번호 | 바꾸지 않을 값 묶음(좌표) |
| 딕셔너리 `dict` | `{키: 값}` | 키 | 가능 | 이름 | 항목 이름이 있는 기록 |
| 집합 `set` | `{ , }` | 없음 | 가능 | 꺼내지 않음 | 중복 제거, 비교 |

"변경 가능"한 자료형은 지난 시간 "변수와 객체"의 원본 보존 문제가 생깁니다.

```python
# [3-4] 집합(set)으로 누락 센서 찾기 — 차집합
installed = {"T01", "T02", "T03", "T04"}    # 설치된 센서
reported = {"T01", "T02", "T04"}            # 오늘 기록을 보낸 센서
print(installed-reported)                 # 기록이 없는 센서
```

## 4. 조건문 · 실습 4 · 슬라이드 12–18
지난 시간에는 조건의 결과를 **출력**했습니다. 오늘부터는 조건의 결과에 따라 **실행할 코드를 고릅니다.** 새로 배우는 것은 `if 조건:` 뒤의 **콜론**과 다음 줄의 **들여쓰기**뿐입니다. 비교 · `and` · `or` · `not`은 지난 시간 그대로입니다.

```python
# [4-1] 지난 시간: 조건의 결과를 출력
t = 29.1
rh = 85
vent = t >= 28 or rh >= 85
print(vent)
```

```python
# [4-2] 오늘: 조건이 참일 때만 실행 — if 조건:  다음 줄은 들여쓰기
if t >= 28 or rh >= 85:
    print("환기 필요")
print("판단 끝")        # 들여쓰지 않은 줄은 조건과 상관없이 실행된다
```

**예상해 보기** [4-1]에서 `t = 24.0`, `rh = 60`으로 바꾸고 [4-1] → [4-2]를 실행하면 무엇이 출력될까요?

```python
# [4-3] 들여쓰기가 없으면 오류 → 앞의 #을 지우고 실행해 보기 (IndentationError)
if t >= 28:
    print("환기 필요")
print("판단 끝")
```

```python
# [4-4] 들여쓰기가 빠진 줄은 블록 밖 — 오류는 없지만 뜻이 달라진다
t = 24.0
if t >= 28:
    print("환기 필요")
    print("aksnjbdju")
print("환기 켬")        # 조건과 상관없이 실행된다 → 들여쓰면 블록 안
```

### 심화 주의 — 들여쓰기 폭
- 한 블록 안에서는 들여쓰기 폭이 같아야 합니다. 코랩은 기본 공백 2칸, 파이썬 표준은 4칸입니다. 섞지만 않으면 됩니다.
- 조건문 안의 `=`는 대입, `==`는 비교입니다. `if t = 28:`은 문법 오류(SyntaxError)입니다.

### if–else · 슬라이드 14
"현재 온도가 기준 이상이라면 환기 필요, 그렇지 않다면 정상"을 그대로 코드로 씁니다. 둘 중 하나만 실행됩니다.

```python
# [4-5] if–else: 둘 중 하나
t = 24.0
t_set = 28.0           # 기준 온도
if t > t_set:
    result = "환기 필요"
else:
    result = "정상"
print(result)
```

### elif와 조건의 순서 · 슬라이드 15
조건이 셋 이상이면 `elif`를 씁니다. **위에서부터 검사하고, 처음 참인 블록 하나만 실행됩니다.**

```python
# [4-6] elif: 위에서부터 검사, 처음 참인 블록 하나만 실행
t = 36.2
if t >= 35:
    state = "고온 경고"
elif t >= 28:
    state = "환기 필요"
else:
    state = "정상"
print(state)
```

**예상해 보기** `t = 29.0`이면? `t = 20.0`이면? 36.2는 `t >= 28`도 참인데 왜 "환기 필요"가 아닐까요?

```python
# [4-7] 순서를 바꾸면: 36.2인데 고온 경고가 나오지 않는다
t = 36.2
if t >= 28:
    state = "환기 필요"
elif t >= 35:
    state = "고온 경고"      # 여기에는 절대 오지 못한다
else:
    state = "정상"
print(state)
```

### 조건식 만들기 · 슬라이드 16
비교(`== != < <= > >=`), `and` · `or` · `not`에 더해 **리스트를 조건에 바로** 씁니다. `in`, `not in`.

```python
# [4-8] in · not in: 리스트에 들어 있는가
registered = ["T01", "T02", "T03"]
sid = "T05"
if sid in registered:
    print(sid, "등록된 센서")
else:
    print(sid, "확인 필요")
```

```python
# [4-9] 범위 조건: 18 이상 28 미만 — 두 비교를 한 번에
t = 24.0
if 18 <= t <28:
    print("적정 범위")
else:
    print("범위 밖")
```

### 판단하기 전에 검사 · 슬라이드 17
센서 값은 비어 있거나(결측) 말이 안 되는 값일 수 있습니다. **판단 조건보다 앞에** 검사 조건을 둡니다.

```python
# [4-10] 결측(None)과 범위 밖을 먼저 걸러낸다
t = None               # 로거가 값을 읽지 못함
if t is None:
    print("결측")
elif t<-30 or t>60:
    print("센서 점검")
elif t >= 28:
    print("환기 필요")
else:
    print("정상")
```

**예상해 보기** `t = 85.0`이면? `t = 29.0`이면?

### 지난 시간에 미룬 두 가지 · 슬라이드 18

```python
# [4-11] 남는 모종이 있으면 트레이 한 판 더
n_plants = 1000
cells = 128
full = n_plants // cells
rest = n_plants % cells
trays = full
if rest > 0:
    trays = full + 1
print(full, rest, trays)
```

```python
# [4-12] 생육도일은 0보다 작으면 0으로
t_max = 12.0
t_min = 4.0
t_base = 10.0
gdd = (t_max + t_min) / 2 - t_base      # -2.0
if gdd < 0:
    gdd = 0
print(gdd)
```

## 5. 종합 · 실습 5 · 슬라이드 20–21
오늘 배운 것을 한 셀에 모읍니다. 딕셔너리에서 값 꺼내기 → 리스트와 `in` → 비교와 `or` → `if / elif / else` → f-string.

**수업에서 다루지 못하면 자유 과제입니다.** 혼자 풀 때는 [3-5], [4-10], [5-1]을 차례로 다시 보면 됩니다.

```python
# [5-1] 종합: 딕셔너리 + 등록 목록 + if–elif–else + f-string
sensor = {"id": "T01", "temp": 29.1, "rh": 85} #creates a dictionary to store data collected from the smart sensor
registered = ["T01", "T02", "T03"] #creates a list containing the IDs of active sensors in the system
if sensor["id"] not in registered:             #python evaluates the cfonditions from top to bottom. Executes the first condition that is True
    result = "센서 확인"
elif sensor["temp"] >= 30 or sensor["rh"] >= 90:
    result = "환경 경고"
elif sensor["temp"] >= 28 or sensor["rh"] >= 85:
    result = "환기 필요"
else:
    result = "정상"
print(f'{sensor["id"]}: {result}')    #use an f-string to dynamically embed variables and dictionary values inside the text string using curly brackets {}
```

### 실습 5 — 기록 한 줄에서 판단까지
`line = "  T03,29.1,85\n"`과 `registered = ["T01", "T02", "T03"]`에 대해
1. 문자열을 정리해 딕셔너리 `rec`로 만든다 (실습 3)
2. 미등록 → `"센서 확인"`, 온도가 −30 미만이거나 60 초과 → `"센서 점검"`, 그다음 [5-1]의 기준대로 판정한다
3. `T03: 환기 필요` 형태로 출력한다

```python
# [5-2] 실습 5
# 1. 정리 → 딕셔너리
line = "  T07,61.5,40\n"
parts = line.strip().split(',')
rec = {"id": parts[0], "temp": float(parts[1]) , "rh": int(parts[2])}
registered = ["T01", "T02", "T03"]
# 2. 검사 → 판정 (검사 조건이 먼저)
if rec ["id"] not in registered:
    result = "센서 확인"
elif rec["temp"]< -30 or rec["temp"] > 60:
    result = "센서 점검"
elif rec["temp"] >= 28 or rec["rh"] >= 85:
    result = "환기 필요"
else:
    result = "정상"
# 3. 출력
print(f'{rec["id"]}: {result}')
```

**바꿔 보기** `line`을 `"  T07,61.5,40\n"`로 바꾸면 어느 조건에서 걸릴까요? 두 조건에 모두 해당하는 값인데 결과는 하나입니다.

출력은: T07: 센서 확인
